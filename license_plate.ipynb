```python
import yaml

data_config ={
    'train':'/content/images/train',
    'val':'/content/images/val',
    'test':'/content/images/test',
    'nc':1,
    'names':['license_plate']
}
```

```python
with open('license_plate_data.yaml', 'w') as f:
    yaml.dump(data_config, f)
```

```python

 from ultralytics import YOLO
 model =YOLO('yolov8n.pt')
 model.train(data='license_plate_data.yaml',epochs=10,
             imgsz=640)
```

```python
from ultralytics import YOLO
import cv2
from google.colab.patches import cv2_imshow
model= YOLO('/content/runs/detect/train-4/weights/best.pt')
results=model.predict(source='/content/images/test',
save=True,conf=0.25)

res_img = results[0].plot()
cv2_imshow(res_img)
```
